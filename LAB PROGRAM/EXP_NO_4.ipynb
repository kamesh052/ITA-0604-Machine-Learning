```python

# Artificial Neural Network using Backpropagation
# Machine Learning Lab Program

import numpy as np

# Step 1: Create the XOR dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

Y = np.array([
    [0],
    [1],
    [1],
    [0]
])

print("Input Dataset:")
print(X)
print("\nTarget Output:")
print(Y)

# Step 2: Initialize network parameters
np.random.seed(42)

input_neurons = 2
hidden_neurons = 4
output_neurons = 1

W1 = np.random.randn(input_neurons, hidden_neurons)
b1 = np.zeros((1, hidden_neurons))

W2 = np.random.randn(hidden_neurons, output_neurons)
b2 = np.zeros((1, output_neurons))

learning_rate = 0.1
epochs = 10000

# Step 3: Define the sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -500, 500)))

# Step 4: Train the network using backpropagation
for epoch in range(epochs):

    # Forward propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    predicted_output = sigmoid(final_input)

    # Calculate error
    error = Y - predicted_output

    # Backpropagation: output layer
    delta_output = error * predicted_output * (
        1 - predicted_output
    )

    # Backpropagation: hidden layer
    delta_hidden = np.dot(
        delta_output, W2.T
    ) * hidden_output * (1 - hidden_output)

    # Update weights and biases
    W2 += learning_rate * np.dot(
        hidden_output.T, delta_output
    )
    b2 += learning_rate * np.sum(
        delta_output, axis=0, keepdims=True
    )

    W1 += learning_rate * np.dot(
        X.T, delta_hidden
    )
    b1 += learning_rate * np.sum(
        delta_hidden, axis=0, keepdims=True
    )

    if (epoch + 1) % 2000 == 0:
        loss = np.mean(error ** 2)
        print(f"Epoch {epoch + 1}, MSE: {loss:.6f}")

# Step 5: Test the trained network
hidden_output = sigmoid(np.dot(X, W1) + b1)
predictions = sigmoid(np.dot(hidden_output, W2) + b2)

print("\nFinal Predictions:")
for i in range(len(X)):
    predicted_class = int(predictions[i, 0] >= 0.5)
    print(
        f"Input: {X[i]} | "
        f"Target: {Y[i, 0]} | "
        f"Predicted: {predictions[i, 0]:.4f} | "
        f"Class: {predicted_class}"
    )

# Step 6: Calculate accuracy
predicted_classes = (predictions >= 0.5).astype(int)
accuracy = np.mean(predicted_classes == Y) * 100

print(f"\nTraining Accuracy: {accuracy:.2f}%")
print("ANN training and testing completed.")
```
