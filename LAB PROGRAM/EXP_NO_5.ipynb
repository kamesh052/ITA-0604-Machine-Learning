```python

# Implementation of K-Nearest Neighbours (K-NN)

# Step 1: Import required libraries
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

# Step 2: Create a sample dataset
X = np.array([
    [1, 2],
    [2, 3],
    [3, 3],
    [6, 7],
    [7, 8],
    [8, 8],
    [2, 1],
    [7, 6],
    [3, 2],
    [8, 7]
])

# Class labels: 0 and 1
y = np.array([0, 0, 0, 1, 1, 1, 0, 1, 0, 1])

print("Dataset:")
print("Features:\n", X)
print("Labels:\n", y)

# Step 3: Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Step 4: Create the K-NN classifier
k = 3
model = KNeighborsClassifier(n_neighbors=k)

# Step 5: Train the model
model.fit(X_train, y_train)

# Step 6: Predict test data
y_pred = model.predict(X_test)

print("\nActual Labels:    ", y_test)
print("Predicted Labels: ", y_pred)

# Step 7: Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy:", round(accuracy * 100, 2), "%")

# Step 8: Classify a new sample
new_sample = np.array([[4, 4]])
prediction = model.predict(new_sample)

print("\nNew Sample:", new_sample[0])
print("Predicted Class:", prediction[0])
print("K-NN classification completed successfully.")
```
