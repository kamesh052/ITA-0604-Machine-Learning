```python

# FIND-S Algorithm in Machine Learning

# Step 1: Create the training dataset
import pandas as pd

data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Warm'],
    'Forecast': ['Same', 'Same', 'Change', 'Same'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

print("Training Dataset:")
print(df.to_string(index=False))

# Step 2: Implement the FIND-S algorithm
def find_s(data):
    attributes = data.iloc[:, :-1].values
    target = data.iloc[:, -1].values

    # Initialize the most specific hypothesis
    hypothesis = None

    print("\nStep-by-step Hypothesis Updates:")

    for i in range(len(target)):
        if target[i] == 'Yes':
            if hypothesis is None:
                hypothesis = attributes[i].copy()
                print(f"Example {i+1}: {list(hypothesis)}")
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j] != attributes[i][j]:
                        hypothesis[j] = '?'

                print(f"Example {i+1}: {list(hypothesis)}")

        else:
            print(f"Example {i+1}: Negative example ignored")

    return hypothesis

# Step 3: Find the final hypothesis
final_hypothesis = find_s(df)

# Step 4: Display the result
print("\nFinal Most Specific Hypothesis:")
print(final_hypothesis)
```
